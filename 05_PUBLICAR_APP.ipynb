# Publicar la app en Streamlit Community Cloud

Ahora tu app solo se ve en tu ordenador, cuando la arrancas con `streamlit run app.py`. Si la publicas en Streamlit Community Cloud, tendrá su propia dirección y cualquiera podrá abrirla desde el navegador.

Son 6 pasos.

| Paso | Qué haces | Dónde |
|---|---|---|
| 1 | Subir lo último a GitHub | GitHub Desktop y github.com |
| 2 | Crear la cuenta de Streamlit | share.streamlit.io |
| 3 | Publicar la app | share.streamlit.io |
| 4 | Arreglar los errores, si salen | Tu app publicada |
| 5 | Poner el enlace en el README | VS Code y GitHub Desktop |
| 6 | Dejarla lista para el jueves | Tu app publicada |

## Antes de empezar: qué hace Streamlit Cloud

Streamlit Cloud es un ordenador en internet que hace lo mismo que tú haces en el tuyo. Se baja tu repositorio de GitHub, instala las librerías que pone en `requirements.txt` y ejecuta `streamlit run app.py`.

En ese ordenador no hay nada tuyo, y por eso casi todos los errores vienen de una de estas tres cosas:

1. Solo ve lo que está en GitHub. Si el CSV no está subido, o has cambiado algo y no has hecho commit y push, para él no existe.
2. Solo instala las librerías que pone en `requirements.txt`. Si en `app.py` importas otra, da error.
3. Funciona con Linux, no con Windows ni con Mac. Linux distingue mayúsculas y minúsculas, así que `Datos.csv` y `datos.csv` son dos archivos distintos. Y en las rutas solo vale `/`, no `\`.

La documentación de Streamlit: [cómo publicar una app](https://docs.streamlit.io/deploy/streamlit-community-cloud/deploy-your-app).

## Paso 1. Subir lo último a GitHub

### 1.1 El requirements.txt

Es la lista de librerías que tiene que instalar la nube, una por línea. Va en la misma carpeta que `app.py`.

La plantilla ya trae uno con `pandas`, `matplotlib`, `streamlit` e `ipykernel`. Si en `app.py` no importas nada más, no tienes que tocarlo. Si importas otra librería, o no usaste la plantilla:

1. Abre `requirements.txt` en VS Code. Si no lo tienes, crea un archivo nuevo en la carpeta del proyecto y llámalo `requirements.txt`.
2. Mira los `import` que hay arriba en `app.py`.
3. Escribe cada librería en una línea, con el nombre que usaste en `pip install`. Casi siempre es lo que va detrás de `import` hasta el primer punto: `import matplotlib.pyplot as plt` es `matplotlib`, e `import plotly.express as px`, `plotly`.
4. Guarda el archivo.

Las que vienen con Python, como `os`, `json` o `math`, no se ponen. Así queda el de la plantilla si le añades `plotly`:

```
pandas
matplotlib
streamlit
ipykernel
plotly
```

En la documentación: [las librerías de la app](https://docs.streamlit.io/deploy/streamlit-community-cloud/deploy-your-app/app-dependencies).

### 1.2 Commit y push

1. Guarda todos los archivos en VS Code.
2. Abre GitHub Desktop. A la izquierda salen los archivos que has cambiado.
3. Abajo a la izquierda, escribe qué has cambiado y pulsa "Commit to main".
4. Arriba, pulsa "Push origin".

Si tu repositorio todavía no está en GitHub, arriba sale "Publish repository" en vez de "Push origin". Antes de pulsarlo, quita la marca de "Keep this code private".

### 1.3 Mirarlo en github.com

En GitHub Desktop, menú Repository → View on GitHub. Se abre tu repositorio en el navegador. Mira tres cosas:

1. Al lado del nombre del repositorio pone "Public". Si pone "Private", tu app también sería privada y solo la verías tú. Para cambiarlo, entra en Settings, baja del todo hasta "Change visibility", elige "Change to public" y confirma en las ventanas que salen. Lo explica [la ayuda de GitHub](https://docs.github.com/es/repositories/managing-your-repositorys-settings-and-features/managing-repository-settings/setting-repository-visibility), que está en español.
2. Ves `app.py`, `requirements.txt` y la carpeta `datos`, y dentro de `datos` está tu CSV.
3. No está la carpeta `.venv`.

## Paso 2. Crear la cuenta de Streamlit

1. Entra en [share.streamlit.io](https://share.streamlit.io) y pulsa "Continue to sign-in".
2. Pulsa "Continue with GitHub" y entra con tu cuenta de GitHub, la misma del repositorio. Si GitHub te pide permiso, pulsa el botón verde de "Authorize".
3. Rellena el formulario y pulsa "Continue".
4. Mira arriba a la izquierda. Si al lado de "Workspaces" sale un aviso, pincha en él y después en "Connect GitHub account". En GitHub, pulsa "Authorize streamlit". Sin esto, Streamlit no puede leer tus repositorios.

En la documentación: [crear la cuenta](https://docs.streamlit.io/deploy/streamlit-community-cloud/get-started/create-your-account) y [conectar GitHub](https://docs.streamlit.io/deploy/streamlit-community-cloud/get-started/connect-your-github-account).

## Paso 3. Publicar la app

1. Arriba a la derecha, pulsa "Create app".
2. Te pregunta si ya tienes una app. Pulsa "Yup, I have an app".
3. Rellena el formulario:
   - Repository: elige tu repositorio en la lista. Sale como `tu-usuario/nombre-del-repositorio`.
   - Branch: `main`. Es la que pone en GitHub Desktop, arriba, en "Current branch".
   - Main file path: viene escrito `streamlit_app.py`. Bórralo y escribe `app.py`. Si en GitHub tu `app.py` está dentro de otra carpeta, pon la carpeta delante, por ejemplo `proyecto-m1/app.py`.
   - App URL: el nombre que tendrá la dirección de tu app, por ejemplo `cervezas-2011`. Minúsculas, números y guiones, de 6 caracteres como mínimo. Si ya lo tiene otra persona, te avisa y pruebas otro.
4. "Advanced settings" no lo toques.
5. Pulsa "Deploy". La primera vez tarda unos minutos, porque tiene que instalar las librerías. Mientras tanto, a la derecha va saliendo lo que hace. Cuando acaba, aparece tu app.
6. Copia la dirección del navegador y pégala en el chat.

Si tu repositorio no sale en la lista, pulsa "Paste GitHub URL". Abre tu `app.py` en github.com, copia la dirección del navegador y pégala ahí.

En la documentación: [publicar la app](https://docs.streamlit.io/deploy/streamlit-community-cloud/deploy-your-app/deploy).

## Paso 4. Si sale un error

### 4.1 Dónde está el error

Si algo falla, en la app sale un recuadro rojo que dice "This app has encountered an error". Ahí no está el error entero. Para verlo, pulsa "Manage app", abajo a la derecha de la app. Se abre un panel con todo lo que ha hecho la nube, y el error está al final.

### 4.2 Los errores más normales

| Lo que ves | Qué pasa | Qué hacer |
|---|---|---|
| En el formulario de publicar pone "This file does not exist" | En GitHub no hay ningún archivo con el nombre que has escrito en "Main file path" | Escribe el nombre exacto, `app.py`, con la carpeta delante si está dentro de otra. Mira en github.com que está subido |
| `ModuleNotFoundError: No module named 'plotly'` | Esa librería no está en `requirements.txt` | Añádela a `requirements.txt` (1.1) |
| `Error during processing dependencies!` | En `requirements.txt` hay algo que no se puede instalar: un nombre mal escrito, una librería que ya viene con Python (como `os`, `json` o `math`) o una que solo funciona en Windows (como `pywin32`, que sale si lo has hecho con `pip freeze`) | Vuelve a hacer el `requirements.txt` a mano (1.1) |
| `FileNotFoundError` y la ruta empieza por `C:\` o por `/Users/` | Es la ruta de tu ordenador, y en la nube no existe | En el `read_csv`, deja solo `datos/` y el nombre del archivo: `pd.read_csv("datos/datos_limpios.csv")` |
| `FileNotFoundError` y en la ruta hay `\` | Linux solo entiende `/` | Cambia las `\` de la ruta por `/` |
| `FileNotFoundError` y la ruta parece bien | El CSV no está en GitHub, o no se llama exactamente así. Para Linux, `Datos.csv` y `datos.csv` son distintos | Mira en github.com, dentro de `datos`, que el CSV está y se llama igual, letra por letra. Si no está, mira que no salga en `.gitignore` y haz commit y push |
| `FileNotFoundError` y tu `app.py` está dentro de otra carpeta | La nube arranca la app desde la carpeta principal del repositorio y busca `datos` ahí | Pon `import os` arriba del todo, y en el `read_csv` cambia la ruta por `os.path.join(os.path.dirname(__file__), "datos", "datos_limpios.csv")` |
| `KeyError` con el nombre de una columna | El CSV que hay en GitHub es más viejo que el tuyo y no tiene esa columna | Vuelve a guardar el CSV desde el EDA y haz commit y push |
| GitHub Desktop no deja subir el CSV porque pasa de 100 megas | GitHub no acepta archivos tan grandes | Guarda el CSV solo con las columnas que usa la app. Si sigue siendo grande, guárdalo comprimido: en el `to_csv` del EDA pon `"datos/datos_limpios.csv.gz"`, y en la app lee ese archivo. pandas lo lee igual |
| `This app has gone over its resource limits` | Los datos no caben en la memoria de la nube | Guarda el CSV solo con las columnas que usa la app |
| Se ve la app de antes, sin lo último que has cambiado | No has hecho push | Haz commit y push (1.2) |

### 4.3 Arreglarlo

1. Corrígelo en tu ordenador y comprueba que la app funciona con `streamlit run app.py`.
2. Haz commit y push.
3. Espera un poco. La app publicada se actualiza sola, sin volver a publicarla.

Si has cambiado `requirements.txt` y sigue saliendo el error, pulsa "Manage app", los tres puntos de abajo del panel y "Reboot app".

Si no consigues arreglarlo, copia las últimas líneas del panel de "Manage app" y pégalas en el chat.

En la documentación: [ver el error](https://docs.streamlit.io/deploy/streamlit-community-cloud/manage-your-app#cloud-logs), [las librerías](https://docs.streamlit.io/deploy/streamlit-community-cloud/deploy-your-app/app-dependencies) y [reiniciar la app](https://docs.streamlit.io/deploy/streamlit-community-cloud/manage-your-app/reboot-your-app).

## Paso 5. El enlace en el README

1. En `README.md`, en el apartado "La app", busca la línea `Enlace: (si la has publicado en Streamlit Community Cloud)` y cambia el paréntesis por la dirección de tu app.
2. Si todavía no tienes la captura, haz una de la app publicada y guárdala como `capturas/app.png`.
3. Haz commit y push.

## Paso 6. Dejarla lista para el jueves

- Si nadie abre la app en 12 horas, se duerme. Ábrela 10 minutos antes de presentar. Si sale "Zzzz. This app has gone to sleep due to inactivity", pulsa "Yes, get this app back up!" y espera a que cargue.
- Ese día no cambies nada a última hora. Cada push vuelve a publicar la app, y si algo se rompe, te pilla en plena presentación.
- Arranca también la app en tu ordenador, con `streamlit run app.py`. Si falla internet o la nube, enseñas esa.

En la documentación: [cuándo se duermen las apps](https://docs.streamlit.io/deploy/streamlit-community-cloud/manage-your-app#app-hibernation).
